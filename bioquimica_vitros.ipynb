# Data Cleaning Shift Report



```python
import pandas as pd

# 1. Carregamento dos dados
# Geralmente, exports do Shift utilizam ';' como separador ou ',' dependendo da regionalização
try:
    df = pd.read_csv('/content/drive/MyDrive/workspace/Colab Notebooks/lab/shift/Listagem Completa de Produtos e Itens Shift 26-02-26.csv', sep=';', encoding='latin-1')
except FileNotFoundError:
    print("Erro: Arquivo não encontrado. Verifique o nome do arquivo.")

# 2. Limpeza básica (remover espaços em branco nos nomes das colunas e strings)
df.columns = df.columns.str.strip()
df = df.apply(lambda x: x.str.strip() if x.dtype == "object" else x)

# 3. Definição dos Filtros (Case Sensitive conforme o banco do Shift)
# No SQL o Agrupamento 1 é 'PRIMEIRO_AGRUPAMENTO' e o 2 é 'SEGUNDO_AGRUPAMENTO'
agrupamento_1_alvo = 'BIOQUIMICA'
agrupamento_2_alvo = 'VITROS'

# 4. Aplicação do Filtro
# Utilizamos o operador lógico & (AND)
df_filtrado = df[
    (df['PRIMEIRO_AGRUPAMENTO'] == agrupamento_1_alvo) &
    (df['SEGUNDO_AGRUPAMENTO'] == agrupamento_2_alvo)
].copy()

# 5. Adicionar coluna vazia para auxílio na contagem física (Opcional)
df_filtrado['CONTAGEM_FISICA'] = ""

# 6. Exportação do resultado
df_filtrado.to_csv('lista_contagem_bioquimica_vitros.csv', index=False, sep=';', encoding='utf-8-sig')

print(f"Sucesso! Foram encontrados {len(df_filtrado)} itens para os critérios selecionados.")
```
